```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import input_file_name, lit

VOLUME_PATH = "/Volumes/nyc_taxi/bronze/nyc_taxi_raw"
BRONZE_TABLE = "nyc_taxi.bronze.raw_trips"
```

```python
#Read the Three Monthly Parquet Files from the Volume
df_april = spark.read.parquet(f"{VOLUME_PATH}/2025-04/yellow_tripdata_2025-04.parquet")
df_may = spark.read.parquet(f"{VOLUME_PATH}/2025-05/yellow_tripdata_2025-05.parquet")
df_june = spark.read.parquet(f"{VOLUME_PATH}/2025-06/yellow_tripdata_2025-06.parquet")

#Combine all 3 Months into One DataFrame
df_all = df_april.union(df_may).union(df_june)

#Write to a delta table
df_all.write \
    .format("delta") \
        .mode("overwrite") \
            .saveAsTable(BRONZE_TABLE)

print(f"Bronze Table Created: {BRONZE_TABLE}")
```
