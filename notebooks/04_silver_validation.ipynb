```python
%sql
SELECT COUNT(*) as total_rows
FROM nyc_taxi.silver.trips_clean;
```

```python
%sql
SELECT
    MIN(tpep_pickup_datetime) as earlist,
    MAX(tpep_pickup_datetime) as latest
FROM nyc_taxi.silver.trips_clean;
```

```python
%sql
SELECT
    SUM(CASE WHEN trip_distance <= 0 THEN 1 ELSE 0 END) as zero_distance,
    SUM(CASE WHEN fare_amount <= 0 THEN 1 ELSE 0 END) as negative_fare,
    SUM(CASE WHEN total_amount <= 0 THEN 1 ELSE 0 END) as negative_total,
    SUM(CASE WHEN tpep_dropoff_datetime <= tpep_pickup_datetime THEN 1 ELSE 0 END) as invalid_dates
FROM nyc_taxi.silver.trips_clean
```

```python
%sql
SELECT
    trip_duration_minutes,
    tip_percentage,
    pickup_hour,
    pickup_day_of_week,
    has_valid_passenger_count
FROM nyc_taxi.silver.trips_clean
LIMIT 5;
```

```python
%sql
SELECT
    SUM(CASE WHEN congestion_surcharge IS NULL THEN 1 ELSE 0 END) as null_congestion_surcharge,
    SUM(CASE WHEN Airport_fee IS NULL THEN 1 ELSE 0 END) as null_Airport_fee,
    SUM(CASE WHEN store_and_fwd_flag IS NULL THEN 1 ELSE 0 END ) as null_store_and_fwd_flag
FROM nyc_taxi.silver.trips_clean
```

```python
%sql
SELECT
    SUM(occurrences - 1) AS total_extra_rows
FROM (
    SELECT
        VendorID,
        tpep_pickup_datetime,
        tpep_dropoff_datetime,
        PULocationID,
        DOLocationID,
        fare_amount,
        COUNT(*) AS occurrences
    FROM nyc_taxi.silver.trips_clean
    GROUP BY ALL
) sub
WHERE occurrences > 1;
```
