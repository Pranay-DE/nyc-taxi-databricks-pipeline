```python
%sql
Select 
    date_trunc('month', tpep_pickup_datetime) as month,
    count(*) AS trips
from nyc_taxi.bronze.raw_trips
group by 1
order by 1;
```

```python
%sql
Select
    SUM(CASE WHEN VendorID IS NULL THEN 1 ELSE 0 END) as null_vendorID,
    SUM(CASE WHEN try_cast(tpep_pickup_datetime as timestamp_ntz) IS NULL THEN 1 ELSE 0 END) as null_tpep_pickup_datetime,
    SUM(CASE WHEN tpep_dropoff_datetime IS NULL THEN 1 ELSE 0 END) as null_tpep_dropoff_datetime,
    SUM(CASE WHEN passenger_count IS NULL THEN 1 ELSE 0 END) as null_passenger_count,
    SUM(CASE WHEN trip_distance IS NULL THEN 1 ELSE 0 END) as null_trip_distance,
    SUM(CASE WHEN RatecodeID IS NULL THEN 1 ELSE 0 END) as null_RatecodeID,
    SUM(CASE WHEN store_and_fwd_flag IS NULL or store_and_fwd_flag = "" THEN 1 ELSE 0 END) as null_store_and_fwd_flag,
    SUM(CASE WHEN PULocationID IS NULL THEN 1 ELSE 0 END) as null_PULocationID,
    SUM(CASE WHEN DOLocationID IS NULL THEN 1 ELSE 0 END) as null_DOLocationID,
    SUM(CASE WHEN payment_type IS NULL THEN 1 ELSE 0 END) as null_payment_type,
    SUM(CASE WHEN fare_amount IS NULL THEN 1 ELSE 0 END) as null_fare_amount,
    SUM(CASE WHEN extra IS NULL THEN 1 ELSE 0 END) as null_extra,
    SUM(CASE WHEN mta_tax IS NULL THEN 1 ELSE 0 END) as null_mta_tax,
    SUM(CASE WHEN tip_amount IS NULL THEN 1 ELSE 0 END) as null_tip_amount,
    SUM(CASE WHEN tolls_amount IS NULL THEN 1 ELSE 0 END) as null_tolls_amount,
    SUM(CASE WHEN improvement_surcharge IS NULL THEN 1 ELSE 0 END) as null_improvement_surcharge,
    SUM(CASE WHEN total_amount IS NULL THEN 1 ELSE 0 END) as null_total_amount,
    SUM(CASE WHEN congestion_surcharge IS NULL THEN 1 ELSE 0 END) as null_congestion_surcharge,
    SUM(CASE WHEN Airport_fee IS NULL THEN 1 ELSE 0 END) as null_Airport_fee,
    SUM(CASE WHEN cbd_congestion_fee IS NULL THEN 1 ELSE 0 END) as null_cbd_congestion_fee
FROM nyc_taxi.bronze.raw_trips;
```

```python
%sql
SELECT
    SUM(CASE WHEN tpep_dropoff_datetime <= tpep_pickup_datetime THEN 1 ELSE 0 END) as invalid_dates,
    SUM(CASE WHEN passenger_count <= 0 THEN 1 ELSE 0 END) as zero_or_negative_passengers,
    SUM(CASE WHEN trip_distance <= 0 THEN 1 ELSE 0 END) as zero_or_negative_trip_distance,
    SUM(CASE WHEN fare_amount <= 0 THEN 1 ELSE 0 END) as zero_or_negative_fare_amount,
    SUM(CASE WHEN total_amount <= 0 THEN 1 ELSE 0 END) as zero_or_negative_total_amount
FROM nyc_taxi.bronze.raw_trips;
```

```python
%sql
SELECT
    VendorID,
    tpep_pickup_datetime,
    tpep_dropoff_datetime,
    PULocationID,
    DOLocationID,
    fare_amount,
    COUNT(*) as occurences
FROM nyc_taxi.bronze.raw_trips
GROUP BY ALL
HAVING COUNT(*) > 1
LIMIT 10;
```
